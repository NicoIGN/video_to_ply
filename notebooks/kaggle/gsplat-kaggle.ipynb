```python
%%bash
export ROOT="/kaggle/working" #/content

cat << 'EOF' > $ROOT/config.sh
#!/bin/bash


export INPUT_DIR="/kaggle/input/datasets/devkinocaen/videos"
export ROOT="/kaggle/working"

export ROOTDIR="$ROOT/evier"
export VIDEOSOURCE="$INPUT_DIR/IMG_4804.MOV"
export INPUT_MODE="video"

#export NUM_FRAMES=150
export FPS=5

#branche dev
export GIT_BRANCH="dev"
export BASENAME="evier"

export PROFILE="gpu/quality"
#export PROFILE="gpu/balanced"
#export PROFILE="cpu/fast"
EOF

cat $ROOT/config.sh
```

```python
%%bash
set -e

INSTALL_DIR="/kaggle/working/miniforge"
INSTALLER="Miniforge3-Linux-x86_64.sh"

# 1) nettoyage si installation cassée ou partielle
if [ -d "$INSTALL_DIR" ]; then
    echo "🧹 Removing existing Miniforge install..."
    rm -rf "$INSTALL_DIR"
fi

# 2) download installer si nécessaire
if [ ! -f "$INSTALLER" ]; then
    echo "⬇️ Downloading Miniforge..."
    wget -q https://github.com/conda-forge/miniforge/releases/latest/download/$INSTALLER
fi

# 3) install propre
echo "⚙️ Installing Miniforge..."
bash $INSTALLER -b -p "$INSTALL_DIR"

# 4) activer conda correctement
source "$INSTALL_DIR/etc/profile.d/conda.sh"

# 5) sanity check
conda --version

# 6) mise a jour config

echo 'export PATH="/kaggle/working/miniforge/bin:$PATH"' >> /kaggle/working/config.sh
```
